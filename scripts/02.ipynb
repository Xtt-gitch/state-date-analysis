```python
import os
import requests
from dotenv import load_dotenv, find_dotenv

load_dotenv(find_dotenv(usecwd=True))

url = "https://open.bigmodel.cn/api/v1/responses"

payload = {
    "model": "glm-5.3",
    "input": "写一首四句的春天七言诗。",
    "stream": True,
}
headers = {
    "Authorization": f"Bearer {os.environ['ZHIPU_API_KEY']}",
    "Content-Type": "application/json",
}

response = requests.post(url, json=payload, headers=headers)

print("HTTP 状态码:", response.status_code)
print(response.text)
```

```python
import os
from dotenv import load_dotenv, find_dotenv
from zai import ZhipuAiClient

# 从当前工作目录逐级向上查找 .env（usecwd=True 使行为不依赖运行方式）
load_dotenv(find_dotenv(usecwd=True))

api_key = os.environ.get("ZHIPU_API_KEY")
if not api_key:
    raise RuntimeError("未找到 ZHIPU_API_KEY，请检查 .env 文件")

# 初始化客户端
client = ZhipuAiClient(api_key=api_key)

# 创建聊天完成请求
response = client.chat.completions.create(
    model="glm-5.3",
    messages=[
        {"role": "system", "content": "您是一个有用的AI助手。"},
        {"role": "user", "content": "您好，请介绍一下自己。"},
    ],
    temperature=0.6,
)

# 获取回复
print(response.choices[0].message.content)
```
